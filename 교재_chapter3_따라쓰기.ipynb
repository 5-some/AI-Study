```python
import pandas as pd

df = pd.read_csv("D:\[실습 파일] 이론 데이터셋\Part2 Titanic.csv")
print(df.head())
```

```python
# 개별 변수의 특징 파악

# 1) 숫자형 변수(Numeric Variable) :

mean_age = df["Age"].mean()
print(f"승객 평균 나이: {mean_age:.2f}세")

median_age =  df["Age"].median()
print(f"승객 나이 중앙값: {median_age:.2f}세")  

max_age = df["Age"].max()
min_age = df["Age"].min()
print(f"최고령 승객: {max_age}세, 최연소 승객: {min_age}세")

total_fare = df["Fare"].sum()
print("총 탑승 요금: {total_fare:.2f}달러")

age_count = df["Age"].count()
print(f"나이가 기록된 승객 수: {age_count}명")




# 2) 범주형 변수(Categorical VAriable) :

# unique() : 특정 열의 고유값 목록을 반환
# nunique() : 특정 열의 고유값 개수를 반환
# value_counts() : 각 값이 "몇 번씩 나왔는지 빈도수"를 세서 반환 = 각 고유값의 개수(빈도)를 계산

# ================= 성별에 따른 승객 수 계산 =================
gender_unique = df["Sex"].unique()
print(gender_unique, "\n")

gender_nunique = df["Sex"].nunique()
print(gender_nunique, "\n")

gender_counts = df["Sex"].value_counts()
print("성별에 따른 승객 수: \n", gender_counts, "\n")


# ================= 객실 등급(Pclass)에 따른 승객 수 계산 =================
pclass_count = df["Pclass"].value_counts()
print("객실 등급에 따른 승객 수:\n", pclass_count, "\n")


# ================= 승선항(Embarked)의 종류 확인 =================
embarked_unique = df["Embarked"].unique()
print("승선항 종류:\n", embarked_unique)
```

```python
# 결측치 탐색: info()와 sinull()

# 결측치(Missing Value) : 값이 비어 있는 것

df.isnull().sum()  # isnull()/isna()를 사용하여 결측치를 확인 => sum()을 적용하면 각 열마다 결측치가 몇 개인지 계산

# 이상치: 데이터의 범위(range)를 확인하기
print(df.describe())
```

```python
# 이상치 탐색:# IQR(사분위수 범위) 방법
 
# 이상치(Outlier) : 다른 값들과 비교했을 때 지나치게 크거나 작은 값

# 1. 사분위수 계산:
Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)
IQR = Q3 - Q1

# 2. 이상치 경계 설정:
lower_bound = Q1 - IQR * 1.5
upper_bound = Q3 + IQR * 1.5

print(f"Q1: {Q1}, Q3: {Q3}, IQR: {IQR}")
print(f"이상치 최소 경계: {lower_bound:.2f}")
print(f"이상치 최대 경계: {upper_bound:.2f}")

# 3. 경계를 벗어나는 이상치 데이터 확인:
outliers = df[(df["Fare"] < lower_bound) | (df["Fare"] > upper_bound)]
print(f"\n총 {len(outliers)}개의 이상치 발견")
print(outliers.head())
```

```python
# 결측치 처리: 삭제 dropna() / 대체 fillna()



# dropna() : 결측치가 포함된 행을 통째로 삭제 => 데이터가 충분히 많고, 결측치가 전체 데이터에서 차지하는 비율이 낮을 때(5% 미만) 사용.

# Titanic 데이터에서 Age에 결측치가 있는 행 확인 

print("처리 전 데이터 수:", len(df))
print("Age 결측치 수:", df["Age"].isnull().sum())

# 결측치가 있는 행 삭제

df_dropped = df.dropna(subset=["Age"])
print("처리 후 데이터 수:", len(df_dropped))



# fillna() : 원하는 값으로 대체

# Age 컬럼의 결측치를 평균(mean)으로 대체
df["Age"] = df["Age"].fillna(df["Age"].mean())
print("Age 결측치 수:", df["Age"].isnull().sum())

# 중앙값으로 대체
df["Age"] = df["Age"].fillna(df["Age"].median)
```

```python
# 이상치 해결: 삭제 / 대체 / 유지



# 1. 이상치 삭제 (Boolean Indexing 사용): 명백한 오류일 때
Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

# 이상치를 제외한 데이터만 필터링
df_no_outlier = df[(df["Fare"] >= lower) & (df["Fare"] <= upper)]
print(f"원본: {len(df)}행 => 이상치 제거 후: {len(df_no_outlier)}행")




# 2. 이상치 대체 (clip()) : 데이터를 잃지 않고 보존할 때.   특정 범위를 벗어나는 값을 경계값으로 대체하는 방법.

# -1- 상한만 설정하기: 300보다 큰 값은 모두 300으로 대체

# df["Speed_Per_Hour"] = df["Speed_Per_Hour"].clip(upper=300)  
# print("처리 후 최대값:", df["Speed_Per_Hour"].max())

# -2- 하한과 상한을 동시에 설정하기 : 10보다 작은 값을 10으로, 300을 초과하는 값을 300으로 대체

# df["Speed_Per_Hour"] = df["Speed_Per_Hour"].clip(lower=10, upper=300)

# -3- IQR 기반으로 자동 계산하여 clip하는 방법
Q1 = df["Fare"].quantile(0.25)
Q3 =df["Fare"].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df["Fare"] = df["Fare"].clip(lower=lower_bound, upper=upper_bound)
print(f"Fare 범위: {df["Fare"].min():.1f} ~ {df["Fare"].max():.1f}")



# 3. 이상치 유지 : 이상치가 실제 데이터의 특성을 반영하는 경우)ex-vup 고객의 매우 높은 구매 금액)에는 그대로 유지히는 것이 적절.
```

```python
# < 그룹별 데이터 분석 > 
# - 데이터를 의미 있는 그룹으로 묶어, 그 그룹들 사이의 특징을 비교하고 관계를 파악
# - 범주형 데이터를 기준으로 숫자형 데이터의 특징이 어떻게 달라지는지를 분석
# - Grouping(그룹화) : 특정 열의 값이 같은 행들을 하나의 그룹으로 묶는 과정

# 내부적 단계: 분할(Split) => 적용(Apply) => 결합(Combine)

# Pclass(1, 2, 3)에 따른 생존율(Survived의 평균)이 출력됨
print(df.groupby("Pclass")["Survived"].mean()) # Pclass를 기준으로 데이터를 그룹화한 뒤, 각 그룹의 "Survived"열의 평균을 계산

# 성별(Sex)에 따른 생존율
df.groupby("Sex")["Survived"].mean()
```

```python
# (Pivot Table)피벗 테이블:
# 많은 데이터를 '가로축(열)'과 '세로축(행)' 두 방향으로 재배치하여 한눈에 보기 쉬운 2차원 교차 요약표로 만들어주는 도구

# groupby: 결과를 1차원(길다란 목록 형태)으로 보여줌
# pivot_table: 결과를 2차원(가로x세로 바둑판 표 형태)으로 보여줌

# 원본 데이터:
┌───┬────────┬──────────┬──────────┐
│   │  Sex   │  Pclass  │ Survived │
├───┼────────┼──────────┼──────────┤
│ 0 │  male  │    3     │    0     │
│ 1 │ female │    1     │    1     │
│ 2 │ female │    3     │    1     │
│ 3 │ female │    1     │    1     │
│ 4 │  male  │    3     │    0     │
└───┴────────┴──────────┴──────────┘

# pivot_table: 내가 지정한 기준에 따라 데이터를 가로와 세로에 나누어 배치하고, 가운데 칸에 평균값을 채워 넣는다.

가로축 (columns = 'Pclass')
                      1등석        2등석        3등석
                 ┌────────────┬────────────┬────────────┐
 세로축   female  │  0.968     │  0.921     │  0.500     │  ◄─ 여성 1,2,3등석 생존율
(index   ────────┼────────────┼────────────┼────────────┤
='Sex')   male   │  0.368     │  0.157     │  0.135     │  ◄─ 남성 1,2,3등석 생존율
                 └────────────┴────────────┴────────────┘
                                   ▲
                         가운데 계산할 값
                        (values = 'Survived')

# Pivot Table을 만드는 코드와 각 매개변수(Parameter)의 의미:
df.pivot_table(
    index='Sex',        # 세로축(행)에 놓을 컬럼
    columns='Pclass',   # 가로축(열)에 놓을 컬럼
    values='Survived',  # 안쪽 칸에 들어갈 계산 대상 데이터
    aggfunc='mean'      # 계산 방법 ('mean': 평균, 'sum': 합계, 'count': 개수 등)
)


# 객실 등급(Pclass)과 성별(Sex)에 따른 생존율을 pivot_table()을 이용해 표현하기:
df.pivot_table(
    index = "Pclass",
    columns = "Sex",
    values = "Survived",
    aggfunc = "mean"
)
```

```python
#  상관관계(Correlation): 한 변수가 변할 때, 다른 변수가 어떻게 함께 변하는지를 나타내는 지표 (ex. 날씨가 더워질수록 아이스크림 판매량이 늘어난다)

# 1. Positive Correlation: 한 변수가 증가할 때, 다른 변수도 함께 증가하는 경향
# 2. Negative Correaltion: 한 변수가 증가할 때, 다른 변수는 오히려 감소가는 경향
# 3. No Correlation: 두 변수 사이에 뚜렷한 관계가 없는 경우

# 상관계수(Correlation Coeffiicient): 관계의 방향과 강도를 -1에서 +1 사이의 숫자로 명확하게 표현한 것
# +1에 가까울수록: 매우 강한 양의 상관관계 / -1에 가까울수록: 매우 강한 음의 상관관계 / 0에 가까울수록: 두 변수 사이에 선형적인 관계가 거의 없음

# .corr() 함수 => 모든 숫자형 변수들 간의 상관관계를 계산 => 표(행렬) 형태로 반환

corr_matrix = df.corr(numeric_only=True)
print(corr_matrix)
```
