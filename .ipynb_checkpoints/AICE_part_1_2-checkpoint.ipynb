```python
import numpy as np

scores_list = [95, 88, 100]

scores_array = np.array(scores_list)

print("파이썬 리스트:", scores_list)
print("Numpy 배열:", scores_array)
print("배열의 타입:", type(scores_array))
```

```python
scores_list= [70, 85, 92, 68]
updated_scores = []

for score in scores_list:
    updated_scores.append(score + 5)

print(updated_scores)
```

```python
import numpy as np

scores_array = np.array([70, 85, 92, 68])
updated_scores = scores_array + 5

print(updated_scores)
```

```python
import numpy as np

scores = np.array([88, 92, 75, 99, 81])

first_score = scores[0]
fourth_score = scores[3]

print("전체 점수:", scores)
print("첫 번째 점수:", first_score)
print("네 번째 점수:", fourth_score)
```

```python
import numpy as np

scores = np.array([88, 92, 75, 99, 81])

middle_scores = scores[1:4]
first_half = scores[:3]
second_half = scores[2:]

print("1번부터 3번까지:", middle_scores)
print("처음부터 2번까지:", first_half)
print("2번부터 끝까지:", second_half)
```

```python
import pandas as pd
import numpy as np

scores_np = np.array([95, 88, 100])
scores_pd = pd.Series([95, 88, 100], index = ['Alice', 'Bob', 'Charlie'])

print("---NumPy Array---")
print(scores_np)
print("\n---Pandas Series---")
print(scores_pd)
```

```python
import pandas as pd

student_data = {
    '국어' : [95, 100, 85],
    '영어' : [85, 95, 90],
    '수학' : [100, 80, 95]
}

df = pd.DataFrame(student_data, index = ['Alice', 'Bob', 'Charlie'])

print(df)

# =========================================================================

math_scores =df['수학']
print("\n---수학 점수---")
print(math_scores)

# =========================================================================

kor_eng_scores = df[['국어', '영어']]
print("\n--- 국어, 영어 점수 ---")
print(kor_eng_scores)

# =========================================================================

second_student = df.iloc[1]

alice_math_score = df.iloc[0, 2]

print("\n--- 1번 위치 학생의 점수---")
print(second_student)
print("\n[0,2] 위치의 값:", alice_math_score)
```
